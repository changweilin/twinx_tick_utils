```python
import matplotlib.pyplot as plt
import numpy as np
from twinx_tick_utils import align_twinx_ticks

# 產生測試資料
x = np.linspace(0, 10, 100)
y1 = np.sin(x) * 3           # 主軸數據
y2 = np.cos(x) * 80          # 副軸數據

# 繪圖
fig, ax1 = plt.subplots(figsize=(8, 4))
ax2 = ax1.twinx()

ax1.plot(x, y1, color='tab:blue', label='sin(x)*3')
ax2.plot(x, y2, color='tab:orange', label='cos(x)*80')

# 設定標籤
ax1.set_ylabel("Left Axis", color='tab:blue')
ax2.set_ylabel("Right Axis", color='tab:orange')
ax1.set_xlabel("X Axis")

# 對齊格線
align_twinx_ticks(ax1, ax2, target_num_ticks=7, prefer='max')

# 顯示圖例
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

# 顯示圖形
plt.title("示範對齊雙 Y 軸格線")
plt.grid(True)
plt.tight_layout()
plt.show()
```
